```python
import pandas as pd
from random_functions import return_dataPIQA, return_dataMGSM

df_arab_standard = return_dataPIQA("arb_arab")
df_farsi = return_dataPIQA("pes_arab")
df_indonesian = return_dataPIQA("ind_latn")
df_javanese = return_dataPIQA("jav_latn")
df_english = return_dataPIQA("eng_latn")
# df_MGSG_INDO = return_dataMGSM("en")

print(f"Saved {len(df_arab_standard)} rows to df_arab_standard.csv")
print(f"Saved {len(df_farsi)} rows to df_farsi.csv")
print(f"Saved {len(df_indonesian)} rows to df_indonesian.csv")
print(f"Saved {len(df_javanese)} rows to df_javanese.csv")
print(f"Saved {len(df_english)} rows to df_english.csv")
```

```python
random_number = [1, 3,67,50,53, 42]
```

```python
import json
from pathlib import Path

language_files = {
    "English": df_english,
    "Indonesian": df_indonesian,
    "Arabic": df_arab_standard,
    "Farsi": df_farsi,
    "Javanese": df_javanese,
    
}

physical_reasoning_questions = []
for number in random_number:
    questions = []
    for language, dataframe in language_files.items():
        record = dataframe.loc[number]
        questions.append({
            "language": language,
            "question": record["prompt"],
            "options": [
                f"{option_index + 1}. {record[f'solution{option_index}']}"
                for option_index in range(4)
            ],
            "answer": int(record["label"]) + 1,
        })
    physical_reasoning_questions.append({"row": number, "questions": questions})

output_path = Path("data/questions.json")
with output_path.open(encoding="utf-8") as file:
    question_data = json.load(file)
question_data["physicial_reasoningQ"] = physical_reasoning_questions
output_path.write_text(
    json.dumps(question_data, ensure_ascii=False, indent=2),
    encoding="utf-8",
)
print(f"Saved {len(physical_reasoning_questions)} rows to {output_path}")
```
